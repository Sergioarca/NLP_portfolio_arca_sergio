### Local Mode

SPEECH TO TEXT

```python
def local_transcript(audio):
    print("Transcribing in locally...")

    model = whisper.load_model("base")
    result = model.transcribe(audio)
    return result['text']

local_transcript(r"C:\Users\sergi\OneDrive\Documentos\UIE\3Curso\2Semestre\Procesamiento del Lenguaje Natural\Exercises\Lab_Exercises_4\audio.mp3")
```

TEXT TO SPEECH

```python
def local_speak(text):
    print(f"Generating local voice: {text}")
    engine = pyttsx3.init('sapi5')

    # RATE
    rate = engine.getProperty('rate')
    print(rate)
    engine.setProperty('rate', 125)

    # VOLUME
    volume = engine.getProperty('volume')
    engine.setProperty('volume',1.0)

    # SAY IT AND FINISH
    engine.say(text)
    engine.runAndWait()
    engine.stop()

local_speak("This is the local text to speech from lab exercise 4")
```

### External API

SPEECH TO TEXT 

```python
def extern_transcript(audio):
    print("Google Api")
    recon = sr.Recognizer()

    with sr.AudioFile(audio) as source:
        audio_data = recon.record(source)
    
    text = recon.recognize_google(audio_data, language="en-US")
    return text

extern_transcript(r"C:\Users\sergi\OneDrive\Documentos\UIE\3Curso\2Semestre\Procesamiento del Lenguaje Natural\Exercises\Lab_Exercises_4\audio_wav.wav")
```

TEXT TO SPEECH

```python
def external_speak(text_gtts):
    print(f"Google API generating voice for :  {text_gtts}")

    # Create the "connect"
    tts = gTTS(text=text_gtts, lang='en')

    audio_archive = "voice_api_external.mp3"
    tts.save(audio_archive)
    print("Starting the generated audio:") # Start the audio
    os.system(f"start {audio_archive}")

external_speak("This is a generated audio by gTTS of google")
```
